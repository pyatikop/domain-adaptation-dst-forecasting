```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns
import numpy as np
import os

import warnings
warnings.filterwarnings('ignore')
```

```python
df_no_adapt = pd.read_excel("../results/результаты_данные_до_2026.xlsx", sheet_name="без адаптации")
df_adapt_d_a = pd.read_excel("../results/результаты_данные_до_2026.xlsx", sheet_name="disc-ace")
df_adapt_a_d = pd.read_excel("../results/результаты_данные_до_2026.xlsx", sheet_name="ace-disc")
```

```python
mask_d_a = ~df_adapt_d_a['adaptation_method'].str.startswith('disc-ace-', na=False)
df_adapt_d_a.loc[mask_d_a, 'adaptation_method'] = 'disc-ace-' + df_adapt_d_a.loc[mask_d_a, 'adaptation_method']

mask_a_d = ~df_adapt_a_d['adaptation_method'].str.startswith('ace-disc-', na=False)
df_adapt_a_d.loc[mask_a_d, 'adaptation_method'] = 'ace-disc-' + df_adapt_a_d.loc[mask_a_d, 'adaptation_method']
```

```python
df = pd.concat([df_no_adapt, df_adapt_d_a, df_adapt_a_d], ignore_index=True)

df['scenario'][df['scenario'] == 'shuffle'] = 'комб.'
df['scenario'][df['scenario'] == 'ace-ace'] = 'A-A'
df['scenario'][df['scenario'] == 'ace-disc'] = 'A-D'
df['scenario'][df['scenario'] == 'disc-ace'] = 'D-A'
df['scenario'][df['scenario'] == 'disc-disc'] = 'D-D'

df['forecast_model'][df['forecast_model'] == 'Linear'] = 'ЛР'
df['forecast_model'][df['forecast_model'] == 'Ridge'] = 'ЛР L1'
df['forecast_model'][df['forecast_model'] == 'Lasso'] = 'ЛР L2'
df['forecast_model'][df['forecast_model'] == 'LGBM'] = 'ГБ'
df['forecast_model'][df['forecast_model'] == 'MLP'] = 'МСП'

df['adaptation_method'][df['adaptation_method'] == 'disc-ace-linreg'] = 'D-A ЛР'
df['adaptation_method'][df['adaptation_method'] == 'disc-ace-lgbm'] = 'D-A ГБ'
df['adaptation_method'][df['adaptation_method'] == 'disc-ace-pls'] = 'D-A ПЛС'
df['adaptation_method'][df['adaptation_method'] == 'disc-ace-dann'] = 'D-A ДСНС'

df['adaptation_method'][df['adaptation_method'] == 'ace-disc-linreg'] = 'A-D ЛР'
df['adaptation_method'][df['adaptation_method'] == 'ace-disc-lgbm'] = 'A-D ГБ'
df['adaptation_method'][df['adaptation_method'] == 'ace-disc-pls'] = 'A-D ПЛС'
df['adaptation_method'][df['adaptation_method'] == 'ace-disc-dann'] = 'A-D ДСНС'

df['adaptation_method'][df['adaptation_method'] == 'disc-ace-cross-linreg+lgbm'] = 'D-A ЛР+ГБ'
df['adaptation_method'][df['adaptation_method'] == 'disc-ace-cross-linreg+mlp'] = 'D-A ЛР+МСП'

df['adaptation_method'][df['adaptation_method'] == 'ace-disc-cross-lgbm'] = 'A-D ЛР+ГБ'
df['adaptation_method'][df['adaptation_method'] == 'ace-disc-cross-mlp'] = 'A-D ЛР+МСП'

df["adaptation_method"] = df["adaptation_method"].fillna("б/а")
df["scenario"] = df["scenario"].fillna("ДА")
df["scenario_adaptation"] = df["scenario"] + " " + df["adaptation_method"]

df['scenario_adaptation'].unique(), df['forecast_model'].unique()
# df[df['adaptation_method'] == 'ace-disc-dann']
```

```python
df.loc[df['delays'] == '24', 'delays'] = '24h'
df.loc[df['scenario_adaptation'] == 'ДА disc-ace-ace-disc-dann', 'scenario_adaptation'] = 'D-A ДСНС'
```

```python
df['scenario_adaptation'].unique()
```

```python
mpl.rcParams['font.size'] = 17         # базовый размер
mpl.rcParams['axes.titlesize'] = 20     # заголовки
mpl.rcParams['axes.labelsize'] = 18     # подписи осей
mpl.rcParams['xtick.labelsize'] = 15    # подписи по осям
mpl.rcParams['ytick.labelsize'] = 15
mpl.rcParams['legend.fontsize'] = 14
```

```python
df_copy = df.copy()

# df_copy = df_copy[df_copy['test_period'] == '2022-2026']
# df_copy = df_copy[df_copy['test_period'] == '2024-2026']
df_copy = df_copy[df_copy['test_period'] == '2022-2024']

hs = [1, 3, 6]

for h in hs:
    target_name = f"Dst_plus{h}"
    # delays = '24h'
    delays = 'auto_func'
    
    df_for_best = df_copy[(df["target"] == target_name) &
                  (df_copy["delays"] == delays)]
    
    df_best = df_for_best.loc[df_for_best.groupby('scenario_adaptation')['R2'].idxmax()]
    df_best = df_best.sort_values('R2', ascending=True)
    
    # df_best = df_best.iloc[-6:] # топ-6 лучших

    test_period = df_best['test_period'].iloc[0]
    
    plt.figure(figsize=(16, 8))
    
    mapping = {"24h": "24ч", "auto_func": "АФ"}
    df_best["delays"] = df_best["delays"].replace(mapping)
    
    y = df_best['scenario_adaptation']
    x = df_best['R2'].values
    xerr = df_best['R2_std'].values
    
    bars = plt.barh(
        y,
        x,
        xerr=xerr,
        linewidth=1.5,
        alpha=0.8,
        ecolor='black',
        capsize=4
    )
    
    if delays == '24h':
        plt.title(f"R² моделей для прогнозов на {h} ч при погружении 24ч на тестовом наборе {test_period}")
    if delays == 'auto_func':
        plt.title(f"R² моделей для прогнозов на {h} ч при погружении по автокорреляционной ф-ции на тестовом наборе {test_period}")
        
    plt.xlabel("R²")
    plt.ylabel("Сценарий обучения / тип адаптации")
    plt.yticks(fontsize=10)
    
    # подписи R² справа от столбцов
    for bar, r2 in zip(bars, df_best['R2']):
        width = bar.get_width()
        plt.annotate(f"{r2:.4f}",
                    xy=(width + 0.002, bar.get_y() + bar.get_height() / 2),
                    ha='left', va='center',  # слева от текста
                    fontsize=17)
    
    # Названия моделей внутри столбцов
    for bar, model in zip(bars, df_best['forecast_model']):
        plt.annotate(model,
                    xy=(bar.get_width() * 0.99, bar.get_y() + bar.get_height() / 2),
                    ha='left', va='center',
                    fontsize=14, fontweight='bold',
                    color='white')
        
    if test_period == '2022-2026':
        if delays == '24h':
            if h == 1:
                plt.xlim(0.85, 0.985)
            if h == 3:
                plt.xlim(0.78, 0.92)
            if h == 6:
                plt.xlim(0.64, 0.77)
        
        if delays == 'auto_func':
            if h == 1:
                plt.xlim(0.83, 0.986)
            if h == 3:
                plt.xlim(0.76, 0.92)
            if h == 6:
                plt.xlim(0.62, 0.77)

    elif test_period == '2024-2026':
        if delays == '24h':
            if h == 1:
                plt.xlim(0.91, 0.985)
            if h == 3:
                plt.xlim(0.78, 0.92)
            if h == 6:
                plt.xlim(0.70, 0.77)
        
        if delays == 'auto_func':
            if h == 1:
                plt.xlim(0.91, 0.987)
            if h == 3:
                plt.xlim(0.86, 0.92)
            if h == 6:
                plt.xlim(0.705, 0.786)

    if test_period == '2022-2024':
        if delays == '24h':
            if h == 1:
                plt.xlim(0.91, 0.985)
            if h == 3:
                plt.xlim(0.78, 0.92)
            if h == 6:
                plt.xlim(0.70, 0.77)
        
        if delays == 'auto_func':
            if h == 1:
                plt.xlim(0.91, 0.977)
            if h == 3:
                plt.xlim(0.815, 0.89)
            if h == 6:
                plt.xlim(0.66, 0.735)
    
    plt.grid(axis='x', alpha=0.3)
    plt.tight_layout()

    # сохранение графика
    # fname = f"R2_Dst_plus{h}_{delays}_{test_period}.png"
    # path = os.path.join('../hists', fname)
    # plt.savefig(path, dpi=300, bbox_inches='tight')
    
    plt.show()
    
    print(df_best[['scenario_adaptation', 'R2', 'delays', 'forecast_model']].round(4))
```

```python
df_copy = df.copy()

# df_copy = df_copy[df_copy['test_period'] == '2022-2026']
# df_copy = df_copy[df_copy['test_period'] == '2024-2026']
df_copy = df_copy[df_copy['test_period'] == '2022-2024']

hs = [1, 3, 6]

# одна фигура и три оси в строку
fig, axes = plt.subplots(1, 3, figsize=(38, 8), sharey=False)

for i, h in enumerate(hs):
    ax = axes[i]

    target_name = f"Dst_plus{h}"
    # delays = '24h'
    delays = 'auto_func'
    
    df_for_best = df_copy[(df_copy["target"] == target_name) &
                          (df_copy["delays"] == delays)]
    
    df_best = df_for_best.loc[df_for_best.groupby('scenario_adaptation')["R2"].idxmax()]
    df_best = df_best.sort_values('R2', ascending=True)
    df_best = df_best.iloc[-6:]   # топ-6
    
    test_period = df_best['test_period'].iloc[0]
    
    mapping = {"24h": "24ч", "auto_func": "АФ"}
    df_best["delays"] = df_best["delays"].replace(mapping)
    
    y = df_best['scenario_adaptation']
    x = df_best['R2'].values
    xerr = df_best['R2_std'].values
    
    bars = ax.barh(
        y,
        x,
        xerr=xerr,
        linewidth=1.5,
        alpha=0.8,
        ecolor='black',
        capsize=4
    )

    if delays == '24h':
        ax.set_title(f"Прогноз на {h} ч")
    else:
        ax.set_title(f"Прогноз на {h} ч")
        
    ax.set_xlabel("R²", fontsize=22)
    if i == 0:
        ax.set_ylabel("Сценарий обучения / тип адаптации", fontsize=20)
    else:
        ax.set_ylabel("")
    
    for bar, r2 in zip(bars, df_best['R2']):
        width = bar.get_width()
        ax.annotate(f"{r2:.4f}",
                    xy=(width + 0.0009, bar.get_y() + bar.get_height() / 2),
                    ha='left', va='center',
                    fontsize=22)
    
    for bar, model in zip(bars, df_best['forecast_model']):
        ax.annotate(model,
                    xy=(bar.get_width() * 0.99, bar.get_y() + bar.get_height() / 2),
                    ha='right', va='center',
                    fontsize=12, fontweight='bold',
                    color='white')
    
    if test_period == '2022-2026':
        # if delays == '24h':
        #     if h == 1:
        #         ax.set_xlim(0.85, 0.985)
        #     if h == 3:
        #         ax.set_xlim(0.78, 0.92)
        #     if h == 6:
        #         ax.set_xlim(0.64, 0.77)
        
        if delays == 'auto_func':
            if h == 1:
                ax.set_xlim(0.955, 0.978)
            if h == 3:
                ax.set_xlim(0.87, 0.908)
            if h == 6:
                ax.set_xlim(0.73, 0.76)

    elif test_period == '2024-2026':
        # if delays == '24h':
        #     if h == 1:
        #         ax.set_xlim(0.91, 0.985)
        #     if h == 3:
        #         ax.set_xlim(0.78, 0.92)
        #     if h == 6:
        #         ax.set_xlim(0.70, 0.77)
        
        if delays == 'auto_func':
            if h == 1:
                ax.set_xlim(0.965, 0.9825)
            if h == 3:
                ax.set_xlim(0.895, 0.917)
            if h == 6:
                ax.set_xlim(0.76, 0.782)

    if test_period == '2022-2024':
        # if delays == '24h':
        #     if h == 1:
        #         ax.set_xlim(0.91, 0.985)
        #     if h == 3:
        #         ax.set_xlim(0.78, 0.92)
        #     if h == 6:
        #         ax.set_xlim(0.70, 0.77)
        
        if delays == 'auto_func':
            if h == 1:
                ax.set_xlim(0.956, 0.972)
            if h == 3:
                ax.set_xlim(0.86, 0.885)
            if h == 6:
                ax.set_xlim(0.695, 0.73)
    
    ax.grid(axis='x', alpha=0.3)

plt.tight_layout()

if delays == '24h':
    fig.suptitle(f"R² моделей ч при погружении 24ч на тестовом наборе {test_period}",
             fontsize=28, y=1.02)
if delays == 'auto_func':
    fig.suptitle(f"R² моделей при погружении по автокорреляционной ф-ции на тестовом наборе {test_period}",
             fontsize=28, y=1.02)

# сохранить одну общую картинку
fname = f"R2_Dst_plus_1_3_6_{delays}_{test_period}.png"
path = os.path.join('../hists', fname)
fig.savefig(path, dpi=300, bbox_inches='tight')

plt.show()
```
